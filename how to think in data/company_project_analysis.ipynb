```python
import pandas as pd 
import numpy as  np
import matplotlib.pyplot as plt
import seaborn as sns
from ydata_profiling import ProfileReport


df=pd.read_excel('company_data.xlsx')
df.head()
#df.dtypes
```

```python
                               #هنا مكتيه هتعرفني الداتا ايه محتويتها بضبط 
                               # بس الداتا هنا طلعت  مش محتاجه اي كلين ومفيش فيها قيم فاضيه 
profile = ProfileReport(df, title="تقرير البيانات", explorative=True)
# عرض التقرير داخل Jupyter
profile.to_notebook_iframe()
```

```python

#contoso نحاول تعد من اكتر شركه اطلب منها اوردات طلعت 
df['Manufacturer'].value_counts()
```

```python
#   يعني فيه بقا حاجات اكبر محتاجين نفهمها contoso وهنا حبين نشوف ايه اكتر الارباح بس مطلعتش 
mp=df.groupby('Manufacturer')['Profit'].sum().sort_values(ascending=False)
plt.bar(x=mp.index, height=mp.values)

plt.xticks(rotation=70)  # يخلي النص رأسي

plt.xlabel('Manufacturer')
plt.ylabel('Profit')
plt.title('who the best profit')
mp
```

```python

# Fabrikam  هنا برضو حبينا نشوف المبيعات طلعت برضو 
sales=df.groupby('Manufacturer')['Sales'].sum().sort_values(ascending=False)
plt.pie(x =sales ,labels=sales.values)

plt.title('who the best Sales ')
plt.grid()
sales
```

```python
# نجرب نشوف مين  اعلي في السعر واقل برضو عشان نفهم اكتر يمكن نوصل لحاجه 

max_val=df.groupby('Manufacturer')['Price'].max().sort_values(ascending=False)
min_val=df.groupby('Manufacturer')['Price'].min().sort_values(ascending=False)

print(max_val)
print("--------------------------------------")
print(min_val)

```
