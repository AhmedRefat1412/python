seaborn
-------

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns


flights= sns.load_dataset("flights")
flights[flights['month']== 'Feb']

# lineplot
plt.figure(figsize=(10,7))
sns.lineplot(x= 'year' ,y= 'passengers' , data= flights, hue='month',size='month')
plt.title(' years vs passengers ')
```

```python
fmri= sns.load_dataset("fmri")
fmri

sns.lineplot(data=fmri, 
             x="timepoint",
             y="signal",
             hue="event" ,# هسيتخدم اللون عشان يفرق بين الايفنتس المختلفه 
             style="region", # سمك الخط هيكون زي ال region 
             )
```

relplot(line plot)
------

```python
fmri= sns.load_dataset("fmri")
fmri

sns.relplot(data=fmri, 
             x="timepoint",
             y="signal",
             hue="event" , 
             style="region", 
             kind='line' ,# there scatter also 
             col='event', 
             row='region'
             )
```

regression plots     (lmplot)
------------------

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np 
import seaborn as sns
sns.set_style('dark')  # الخلفيه 

tips = sns.load_dataset('tips')
tips
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
sns.regplot(data=tips,x='size',y='tip',x_jitter=0.1)  # لو في قيم كتير في نفس المكان ممكن بوضحها 
```

```python
# hue -row -col ان ال ام بلوتس اكثر مرونه وفيها   regplots and lmplots الفرق بين  
tips
sns.lmplot(data=tips,
           x='size',
           y='tip',
           hue='smoker', 
           col='smoker',
           row='sex',
           truncate=False, # دا بيخلي الخط الي مرسوم او الميل بيبقا كامل مش من بدايه الداتا لاخرها 
           
           ) 
```
